```python
df_time_features.iloc[:,0:].std()
```

```python
time_domain_features = pd.DataFrame(df_time_features.iloc[:,0:].std())
```

```python
bs_time_domain_features_std=[]
bs_time_domain_features_lb=[]
bs_time_domain_features_ub=[]
for i in range(0,len(df_time_features.columns)):
    result=(bs.bootstrap(df_time_features.iloc[:,i].to_numpy(),stat_func=bs_stats.std))
    bs_time_domain_features_std.append(result.value)
    bs_time_domain_features_lb.append(result.lower_bound)
    bs_time_domain_features_ub.append(result.upper_bound)

time_domain_features.insert(1,'bootstrap_value',bs_time_domain_features_std)
time_domain_features.insert(2,'bootstrap_lower_bound',bs_time_domain_features_lb)
time_domain_features.insert(3,'bootstrap_upper_bound',bs_time_domain_features_ub)
```

# Q2

ISLR 3.7.4

1. The training RSS will be less for Cubic regression as there are more parameters involved i.e., more degrees of freedom, which can try to overfit.

2. If the true relation between X and Y is linear then test RSS for the linear regression will be less, as the values predicted for any input will be close to the actual value.

3. The training RSS will be less for Cubic regression as there are more parameters involved i.e., more degrees of freedom, which can try to overfit.

4. If the true relation between X and Y is not linear then test RSS for the cubic regression will be less, as the values predicted for any input will be close to the actual value which will be having a non-linear relation with the parameters.
